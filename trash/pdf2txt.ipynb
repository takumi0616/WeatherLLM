```python
from PyPDF2 import PdfReader

def extract_text_from_pdf_with_pypdf2(pdf_path):
    text = ''
    with open(pdf_path, 'rb') as file:
        pdf = PdfReader(file)
        for i in range(len(pdf.pages)):
            page = pdf.pages[i]
            text += page.extract_text()
    return text
```

```python
def text_conv(txt,ym):
  txt_output = ym
  for i,t in enumerate(txt.split('\n')):
    if i == 0:
      txt_output += t.split('日')[0].zfill(2) + ','
      s = t.split(')')
      txt_output += s[0] + '),' + s[1]
      txt_output += ','
    else:
      txt_output += t
  return txt_output
```

```python
def text_write(pdf_path,txt):
  ym = '20' + pdf_path.split('/')[-1].replace('.pdf','')
  f = open(pdf_path.replace('pdf','csv'),'w')
  txts = []
  i = txt.find('日(')
  start = 0
  while i > 0:
    for j in range(i - 1,0,-1):
      if txt[j].isdigit() == False:
        end = j + 1
        break
    if start != 0:
      f.write(text_conv(txt[start:end],ym) + '\n')
    start = end
    i += 1
    i = txt.find('日(',i)
  f.write(text_conv(txt[start:],ym))
  f.close()
```
